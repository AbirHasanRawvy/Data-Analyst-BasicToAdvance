# **Missing Value Handle**

# **{"Key": ["value", "value", "value"]}**

```python
dataset = {"Name" : ["Ratul", "Fahmid", "Tahfim", "Atik"],
           "Age" : [25, np.nan, 25, np.nan],
           "Salary" : [50000, 35000, np.nan, 44000],
           "Gender" : ['M','M', np.nan, 'M']
           }
```

```python
dataset
```

# **Table Format**

```python
df = pd.DataFrame(dataset)
df
```

```python
df.isnull().sum() #check missing value
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Age']
```

```python
df
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df
```

# **Group By Analysis**

```python
sales = pd.DataFrame({
    "City" : ["Dhaka", "Dhaka", "Dhaka", "Dhaka", "Dhaka"],
    "Category" : ['Laptop', 'Mouse', 'Mouse', 'Laptop', 'Laptop'],
    "Sales" : [20000, 4000, 3000, 30000, 50000]
})
```

```python
sales
```

```python
sales.groupby('City')['Sales'].mean()
```
