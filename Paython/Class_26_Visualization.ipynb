# **Heatmap**

```python
import pandas as pd #Libraries;

df = pd.read_csv("/content/DataSheet.csv")
print(df)
df.head() # first 5 rows
df.info() # data types & null
df.describe() # Summary of numerical data
df.shape # rows & columns
df.columns # column names

```

# **Missing value handaling**

```python
df.isnull().sum() # check missing value.
```

# **Detect duplicates value**

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

# **Data visualization **

```python
sns.histplot(df['Amout'], bins=10, kde=True, color='skyblue')
plt.xlabel('Amount')
plt.ylabel('Frequency')
plt.title('Amount visualization')
plt.show()
```

# **Categorical Colum**

```python
df['ProductName'].value_counts()
```

# **count 0 & 1. Furthermore visualization**

```python
sns.boxplot(df['Amout'])
plt.xlabel('Amount')
plt.ylabel('Frequency')
plt.title('Amount visualization')
plt.show()
```

```python
Corr = df.corr(numeric_only=True)

sns.heatmap(Corr, annot=True, cmap='coolwarm')
plt.title('Correlation Between Numerical Deatures')
plt.show()
```
