```python
import matplotlib.pyplot as plt
```

```python
from matplotlib.lines import lineStyles
x=[1, 2, 3, 4, 5]
y=[10, 20, 30, 40, 50]

plt.plot(x,y, color="blue", marker="s", linestyle="--", linewidth="3")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Line Plot")
plt.show()
```

# **Scater Plot**

```python
z=[2, 4, 6, 8, 10]
u=[9, 18, 27, 36, 45]
plt.scatter(z,u, color='red', marker = 's')
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.title("Scater")
```
